# Level 2: Customer churn analysis

```python
# 11. How many customers have churned and not churned?

df['churned'].value_counts()
```

```python
# churned value
(df['churned'] == 'Yes').value_counts()
```

```python
# 12. What is the overall customer churn rate?
churn_rate = (df['churned'] == 'Yes').mean()*100
not_churn_rate = (df['churned'] == 'No').mean()*100
print(f"Churned rate: {churn_rate:.2f}%")
print(f"Not churned rate: {not_churn_rate:.2f}%")
```

```python
# 13. How many customers are still active?
print("customers are still active", (df['churned'] == 'No').sum())
```

```python
# 14. percentage of customer retain

not_churn_rate = (df['churned'] == 'No').mean()*100
print(f"Not churned rate: {not_churn_rate:.2f}%")
```

```python
# 15. What is the churn rate by subscription type?
churn_rate_subscription_type = df.groupby('subscription_type').agg(
    Total_Customer = ('churned', 'size'),
    churned_customer = ('churned', lambda x: (x == 'Yes').sum())
)

churn_rate_subscription_type['churned'] = (
    (churn_rate_subscription_type['churned_customer'] / 
     churn_rate_subscription_type['Total_Customer'])*100
).round(2)

print(churn_rate_subscription_type)
```

```python
# 16. Which customer segment has the highest churn rate?
# subscription type, age, region
# customer segment by age

bins = [18,25,50,75]
labels = ['18-24 age', '25-49 age', '50-100 age']

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df1 = df.groupby('age_group')['age'].agg(
    ['count','mean','min','max']
).round(2)

print("Number of age group")
df1
```

```python
# 17. Does churn vary by customer age group?

df2 = df[df['churned'] == 'Yes'].groupby('age_group')['age'].agg(
    ['count','min','max']
)
print()
print('Age group who churned more')
df2
```

```python
# 18. Does churn vary by gender or other demographic groups?
df3 = df[df['churned'] == 'Yes'].groupby('gender')['churned'].agg(
     ['count']
)

print("Churn by gender")
df3
```

```python
df4 = df[df['churned'] == 'Yes'].groupby('region')['churned'].agg(
    ['count']
)

print('churn by region')
df4
```

```python
# 19. Which customer favorite_genre channel has the highest churn?
df5 = df[df['churned'] == 'Yes'].groupby('favorite_genre')['churned'].agg(
    ['count']
)

print('Highest of favorite_genre churned')
df5.idxmax()
```

# Level 3: Revenue and customer behaviour


```python
# 20. What is the average monthly subscription charge?
d1 = df.groupby('subscription_type')['monthly_fee'].agg(
    ['mean']
)
print('Average subscription charge')
d1
```

```python
# 21. Do customers who pay more churn more frequently?
premium_churn = df[
    (df['subscription_type'] == 'Premium') &
    (df['churned'] == 'Yes')
]

print("Churned Premium customers:", len(premium_churn))
```

```python
premium = df[df['subscription_type'] == 'Premium']

premium_churn_rate = (
    (premium['churned'] == 'Yes').mean() * 100
)

print(f"Premium churn rate: {premium_churn_rate:.2f}%")
```

```python
# 22. What is the monthly revenue associated with churned customers?
d1 = df.groupby('churned')['monthly_fee'].sum()
print("monthly revnue by churned and not churned customers ",d1.round(2))
```

```python
# 23. Which subscription plan generates the most revenue?
d2 = df.groupby('subscription_type')['monthly_fee'].sum()
print(f"The subscription plan generates the most revenue is {d2.idxmax()}")
```

```python
# 24. Which customer segment has both high churn and high revenue?
# 1. Analyze by subscription type

segment_analysis = df.groupby('subscription_type').agg(
    Total_Customers=('churned', 'count'),
    Churned_Customers=('churned', lambda x: (x == 'Yes').sum()),
    Total_Revenue=('monthly_fee', 'sum')
)

segment_analysis['Churn_Rate'] = (
    segment_analysis['Churned_Customers'] /
    segment_analysis['Total_Customers'] * 100
).round(2)

segment_analysis['Churned_Revenue'] = df[df['churned'] == 'Yes'].groupby(
    'subscription_type'
)['monthly_fee'].sum()

print(segment_analysis.fillna(0).round(2))
```

```python
# 2. Analyze by age group
age_analysis = df.groupby('age_group', observed=False).agg(
    Total_Customers=('churned', 'count'),
    Churned_Customers=('churned', lambda x: (x == 'Yes').sum()),
    Total_Revenue=('monthly_fee', 'sum')
)

age_analysis['Churn_Rate'] = (
    age_analysis['Churned_Customers'] /
    age_analysis['Total_Customers'] * 100
).round(2)

age_analysis['Churned_Revenue'] = (
    df[df['churned'] == 'Yes']
    .groupby('age_group', observed=False)['monthly_fee']
    .sum()
)

print(age_analysis.fillna(0).round(2))
```
